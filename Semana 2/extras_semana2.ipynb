### Extra 1: Control de aforo con break y continue (while)
Procedimiento: Se desarrolló un código que registrara visitantes y acumulara el costo de sus boletos usando un ciclo while. Se utulizó continue para que se omitiera el cobro de los menores de 3 años y break cuando se alcanzó/superó la meta de 100$.

```python
menor_3 = 0
menor_18 = 30
mayor_edad = 45

tote = 0

while True:
    edad = int (input("Ingrese la edad del visitante: "))
    if edad < 3:
        cost = menor_3
        print("Menor de 3 años. Sin Cargo.")
        continue

    elif edad <18:
        cost = menor_18

    else:
        cost = mayor_edad

    tote = tote + cost

    if tote >= 100:
        print("Aforo máximo alcanzado, el registro se ha detenido.")
        break

print(f"Total Final: ${tote:.2f}$")







```

### Extra 2: Estadística de visitantes con for
Procedimiento: Se escribió un código que registra las edades de un número específico de visitante. Se uso un ciclo for para recolectar la edad de cada visitante, un contador para identificar a los adultos y un acumulador para sumar sus edades y calcular el promedio.


```python
visits = int(input("¿Cuántos visitantes tenemos hoy?"))

oldies = 0
age_sum = 0

for i in range (visits):
    age = int(input(f"Edad Visitante {i+1}: "))

    age_sum = age_sum + age

    if age >= 18:
        oldies = oldies + 1

avg = age_sum / visits

print(f"Adultos: {oldies}")
print(f"Promedio Edad:  {avg:.2f}")
```

### Extra 3: Depuración de un cobro con PDB
Se realizó depuración al código otorgado para calcular el precio final de un objeto. Se utilizó pdb.set_trace como medio de depuración. Posterior a esto se escribió el código correcto, permitiendo así que el cálculo se llevara a cabo.

```python
import pdb

precio = 45
descuento = 12
pdb.set_trace()

total = precio - descuento
print(f"Total: ${total:.2f}")
```

Al utilizar pdb.set_trace, se notó que el error consiste en que el descuento no está marcado como porcentaje, sino como número entero. A continuación, se muestra el código corregido.

```python
import pdb
precio = 45

descuento = precio *  0.12

total = precio - descuento

print(f"Total: ${total:.2f}")
```

### Extra 4: Pirámide de asteriscos con for
Se diseñó y desarrolló código para generar una pirámide hecha de asteriscos, partiendo de la altura indicada. Se utilizaron ciclos for anidados para controlar la cantidad de asteriscos y las filas. Todo esto en conjunto con end="" para que se mantuvieran en la misma fila.


```python
height = int(input("Definamos la altura de la pirámide:"))
for fila in range(1, height + 1):

    for asterisco in range (fila):
        print("*", end="")
    print()
```
