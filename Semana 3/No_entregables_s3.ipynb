Programa que pida 3 números y diga cual es el mayor

```python
numero1= int(input("Dame un número: "))
numero2= int(input("Dame otro número: "))
numero3= int(input("Dame uno más: "))


if numero1>numero2 and numero1>numero3:
    print(f"El número más grande es el: " , (numero1))
elif numero2>numero1 or numero2>numero3:
    print(f"El número más grande es el: " , (numero2))
else:
    print(f"El número más grande es el: " , (numero3))




```

Programa que pide numeros positivos y al ingresar el número 0 se detenga

```python

suma=0
while True:
    sum = float(input("Por favor ingrese números positivos: "))

    if sum == 0:
        break
    else:
        suma += sum
        continue
print(suma)
   
        


```
