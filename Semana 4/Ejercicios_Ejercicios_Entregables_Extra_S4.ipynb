### Este código muestra una matriz y calcula la suma de todos sus valores, para lo cual se utilizan ciclos for

```python
matriz= [[1,2,3], [4,5,6],[7,8,9]]

def imprimir_matriz(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end=" ")
        print()

def suma_matriz(matriz):
    suma=0

    for fila in matriz:
        for elemento in fila:
            suma+=elemento

    return suma

imprimir_matriz(matriz)
print(f"La suma de los valores es {suma_matriz(matriz)}")
```
