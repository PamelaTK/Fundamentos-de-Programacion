PREGUNTAR CUANTAS CALIFS DESEA PROMEDIAR Y PROMEDIARLAS

```python
def promedio(sumatoria, califs):
    return sumatoria / califs

califs = int(input("¿Cuántas califs desea ingresar?"))
sumatoria = 0

print("se van a pedir {califs} calificaciones")
for c in range (1, califs+1):
    calif = float(input(f"dame la calificación {c}"))
    sumatoria = sumatoria = sumatoria + calif

print(f"El promedio de {califs} es {promedio(sumatoria,califs)}")
    
```

```python
lista = []
contador = 1
while contador <= 10:
    valor = int(input("Ingresa un número entre el 10 y -10: "))
    if valor >= -10 and valor <= 10:
        contador += 1
        lista.append(valor)
    else:
        print("Valor no válido, ingrese uno dentro de los lineamientos:")

    print(lista)
    for i in lista:
        if i < 0:
            print("asteriscos negativos")
        elif i>0:
            for a in range(10):
                print("*" , end="")
            print("|" , end="")
            for a in range(i):
                print("*", end="")
            print()
        else:
            print("!")


```
