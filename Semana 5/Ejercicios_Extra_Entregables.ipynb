### Extra 1: Sistema de calificaciones con tuplas
**Enunciado:** Crea una tupla `calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)`.
Imprime la tercera calificación, captura dos calificaciones nuevas con `input()` y
anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala **de mayor
a menor** y muestra la suma de todos los elementos usando una función que retorne
el valor.

```python
califs = (7.5, 9.0, 8.0, 6.5, 10.0)

print("La tercera calificación es:" , califs[2])

ncalif1= int(input("Ingrese una calificación adicional: "))
ncalif2= int(input("Ingrese una más: "))

califsnew= califs + (ncalif1, ncalif2)
lista = list(califsnew)
lista.sort()
print(lista)
print("Las lista con las calificaciones nuevas y ordenadas se ve así:" , lista)

type(print)
```

### Ejercicio 2: Longitud de una tupla
**Enunciado:** Dada la tupla `frutas = ("manzana", "plátano", "cereza",
"durazno")`, calcula e imprime la cantidad de elementos con la función `len()`.
**Entrada:** `frutas = ("manzana", "plátano", "cereza", "durazno")`
**Salida:** `La tupla tiene 4 elementos`

```python
frutas = ("manzana", "plátano", "cereza", "durazno")

frutas.__len__

print("La tupla tiene" , (len(frutas)), "elementos")
```

### Ejercicio 3: Índices negativos
**Enunciado:** Dada la tupla `puntos = (5, 10, 15, 20, 25, 30)`, imprime el último
elemento usando el índice negativo `-1` y el tercero desde el final usando `-3`.
**Entrada:** `puntos = (5, 10, 15, 20, 25, 30)`
**Salida:** `El último elemento es: 30` y `El tercero desde el final es: 20`

```python
puntos = (5,10,15,20,25,30)
print("El último elemento es:" , (puntos[-1]))
print("El tercero desde el final es:" , (puntos[-3]))
```
