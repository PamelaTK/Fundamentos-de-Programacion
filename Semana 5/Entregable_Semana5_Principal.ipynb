```python
def sumar_tupla(numeros):
    return sum(numeros)
def ejecutar_tuplas():
    numeros= (5, 10, 15, 20, 25)

    print("\n Tuplas")
    print("Tercer elemento", numeros[2])

    num1= int(input("Ingrese un número extra: "))
    num2= int(input("Ingrese uno más: "))

    numeros = numeros + (num1, num2)

    print("Nueva Tupla:", numeros)

    lista_nums = list(numeros)
    lista_nums.sort()

    print("Lista Ordenada:", lista_nums)
    print("Total Suma:", sumar_tupla(numeros))


def buscar_fono(contactos, nombre):
    return contactos.get(nombre)

def ejecutar_diccionario():
    contactos={
        "Ana":"555-0101",
        "Luis": "555-0102",
        "Mia":"555-0103"
    }

    print("\n Diccionarios")
    nuevo_nom =input("Ingrese el nombre del nuevo contacto: ")
    nuevo_tel= input("Ingrese el número del nuevo contacto: ")

    contactos[nuevo_nom]= nuevo_tel
    print("\n Nombres de los contactos:")

    for nombre in contactos.keys():
        print(nombre)

    nombre_buscar = input("\n Ingrese el nombre del contacto que desea buscar: ")
    telefono = buscar_fono (contactos, nombre_buscar)

    if telefono:
        print("El telefono de:", nombre_buscar, "es:", telefono)
    else:
        print("Contacto no encontrado.")


def ejecutar_excepciones():
    print("\n Excepciones")

    try:
        num1=int(input("Ingrese un número entero: "))
        num2=int(input("Uno más: "))

        print("Suma: ", num1+num2)

        try:
            resultado =num1/num2
            print("División: ", resultado)

        except ZeroDivisionError:
            print("Error: No se puede dividir entre 0.")

    except ValueError:
        print("Error: Debe ingresar números válidos.")

def contar_palabras(mensaje):
    palabras = mensaje.split()
    return len(palabras)

def ejecutar_strings():
    mensaje="Python es un lenguaje poderoso"

    print("\n Strings")
    print("Mensaje", mensaje)
    print("Longitud:", len(mensaje))
    print("Mayúsculas:", mensaje.upper())
    print("Reemplazo:", mensaje.replace("Python","programación"))
    print("Cantidad de palabras:", contar_palabras(mensaje))

            
while True:
    print("\n Menú Principal")
    print("1. Tuplas")
    print("2. Diccionarios")
    print("3. Excepciones")
    print("4. Strings")
    print("5. Finalizar")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        ejecutar_tuplas()

    elif opcion == "2":
        ejecutar_diccionario()

    elif opcion == "3":
        ejecutar_excepciones()

    elif opcion == "4":
        ejecutar_strings()

    elif opcion == "5":
        print("Programa Finalizado.")
        break

    else:
        print("Opción no válida.")

    #EL código funciona correctamente.#
```
