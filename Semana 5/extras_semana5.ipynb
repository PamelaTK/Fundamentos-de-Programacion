### Extra 1: Sistema de calificaciones con tuplas
**Enunciado:** Crea una tupla `calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)`.
Imprime la tercera calificación, captura dos calificaciones nuevas con `input()` y
anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala **de mayor
a menor** y muestra la suma de todos los elementos usando una función que retorne
el valor.

```python
califs = (7.5, 9.0, 8.0, 6.5, 10.0)

print("La tercera calificación es:" , califs[2])

ncalif1= int(input("Ingrese una calificación adicional: "))
ncalif2= int(input("Ingrese una más: "))

califsnew= califs + (ncalif1, ncalif2)
lista = list(califsnew)
print("La nueva lista es:", lista)
lista.sort(reverse=True)
print("Las lista con las calificaciones nuevas y ordenadas se ve así:" , lista)
```

### Extra 2: Agenda de contactos con búsqueda
**Enunciado:** Crea un diccionario `agenda` con tres contactos iniciales (`"Ana":
"555-0101"`, `"Luis": "555-0102"`, `"Mía": "555-0103"`). Captura un nuevo contacto
(nombre y teléfono) con `input()` y agrégalo al diccionario. Luego imprime todos
los nombres con `keys()` en una sola línea y, con una función que retorne el valor,
busca y muestra el teléfono del nuevo contacto.

```python
def buscar_contacto(agenda,nombre):
    return agenda.get(nombre)

agenda = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"

}

nuevo_contact =input("Ingrese el nombre del contacto que desea agregar: ")
nuevo_tel = input("Ahora ingrese el número de contacto: ")

agenda[nuevo_contact]=nuevo_tel

print("Contactos registrados:", end="")

for nombre in agenda.keys():
    print(nombre,end=",")

nombre_buscar = input("\n Nombre a buscar: ")

telefono = buscar_contacto(agenda, nombre_buscar)

if telefono:
    print("El telefono de", nombre_buscar, "es", telefono)
else:
    print("Contacto no encontrado.")
    
```

### Extra 3: Calculadora segura con manejo de excepciones
**Enunciado:** Escribe un programa que pida dos números enteros. Implementa un
bloque `try-except` que capture `ValueError` (caracteres no numéricos o vacíos) con
un mensaje controlado y un bloque específico `except ZeroDivisionError` que muestre
un mensaje amigable si el segundo número es cero. En caso exitoso, muestra la
división del primero entre el segundo.

```python
try:
    num1 = int(input("Primer número entero: "))
    num2 = int(input("Segundo número entero: "))

    try:
        resultado = num1/num2
        print(num1, "dividido entre", num2, "es:", resultado)

    except ZeroDivisionError:
        print("Error: No es posible dividir entre 0. Ingresa un divisor distinto a 0.")
except ValueError:
    print("Error: Debe ingresar un número entero válido.")
```

### Extra 4: Analizador de mensajes con strings
**Enunciado:** Crea una variable `mensaje = "Python es un lenguaje poderoso"`.
Imprime su longitud con `len()`, conviértelo a mayúsculas, reemplaza la palabra
`"Python"` por `"programación"` y, con una función que retorne el conteo, muestra
cuántas palabras contiene el mensaje original.

```python
def contar_palabras(mensaje):
    palabras = mensaje.split()
    return len(palabras)

mensaje="Python es un lenguaje poderoso"

print("Longitud del mensaje", len(mensaje), "caracteres")
print("En mayúsculas el mensaje se ve así:", mensaje.upper())
print("El texto reemplazado es:", mensaje.replace("Python", "programación"))
print("Palabras totales:", contar_palabras(mensaje))
```

### Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario
el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión
y, al final, imprime con `items()` el color ganador junto con su total.

```python

votos = {
    "ROJO": 0,
    "AZUL": 0,
    "VERDE": 0
}

for i in range(5):
    color= input("Ingrese su voto (ROJO, AZUL, VERDE): ")
    color = color.upper()


    if color in votos:
        votos[color] = votos[color] + 1
    else:
        print("Color no valido.")

ganador= ""
mayor= 0

for color, cantidad in votos.items():
    print(color, ":", cantidad)
    
    if cantidad > mayor:
        mayor = cantidad
        ganador = color

print("El color ganador es:", ganador)
print("Votos Totales:", mayor)
```

### Desafío Extra 3: Validación robusta de entrada
Escribe un programa que pida al usuario dos números enteros dentro de un bloque
`try-except`. Si el usuario ingresa texto o un valor vacío, muestra un mensaje
controlado y repite la solicitud con un ciclo `while`. Luego pide la división del
primero entre el segundo, con un bloque específico para `ZeroDivisionError`.

```python
while True:
    try:
        numero1 = int(input("Primer número entero: "))
        numero2 = int(input("Segundo número entero: "))
        
        try:
            resultado = numero1/numero2
            print(numero1, "dividido entre", numero2, "es:", resultado)
            break
        except ZeroDivisionError:
            print("Error: No es posible dividir entre 0. Ingresa un divisor distinto a 0.")
    except ValueError:
        print("Error: Debe ingresar un número entero válido.")
        continue
    
```

### Desafío Extra 5: Minimenú modular integrador
Construye un menú con un ciclo `while` y opciones numéricas que permita ejecutar de
forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en
el diccionario `contactos`, (3) dividir dos números con manejo de
`ZeroDivisionError` y (4) analizar un mensaje de texto. La opción (5) debe
finalizar el programa.

```python
def mostrar_tupla(numeros):
    lista = list(numeros)
    lista.sort()
    print("Tupla ordenada:", lista)

def buscar_telefono(contactos, nombre):
    return contactos.get(nombre)

def dividir_numeros(num1, num2):
    try:
        resultado =num1/num2
        print("Resultado: ", resultado)

    except ZeroDivisionError:
        print("Error. Ese número no puede dividirse entre 0.")

def contar_palabras(mensaje):
    palabras=mensaje.split()
    return len(palabras)

nums= (5, 10, 15, 20, 25)
contactos={
    "Ana":"555-0101",
    "Luis":"555-0102",
    "Mía": "555-0103"
}

mensaje ="Python es un lenguaje poderoso"

while True: 
    print("\n Menú Modular")
    print("1. Tupla Ordenada:")
    print("2. Buscar Teléfono:")
    print("3. Dividir 2 Números:")
    print("4. Analizar Mensaje:")
    print("5. Finalizar")

    opcion= input("Seleccione una opción (del 1-5)")

    if opcion == "1":
        print(mostrar_tupla(nums))

    elif opcion == "2":
        nombre=input("ingrese el nombre que desea buscar: ")
        telefono= buscar_telefono(contactos, nombre)
        if telefono:
            print("El teléfono:", nombre, "es:", telefono)
        else:
            print("Contacto no encontrado.")

    elif opcion == "3":
        num1= int(input("Ingrese el primer número: "))
        num2= int(input("Ingrese uno más: "))
        dividir_numeros(num1, num2)

    elif opcion == "4":
        print("Mensaje:", mensaje)
        print("Longitud:", len(mensaje))
        print("Mayúsculas:", mensaje.upper())
        print("Reemplazo:", mensaje.replace("Python", "Programación"))
        print("Cantidad de palabras:", contar_palabras(mensaje))

    elif opcion == "5":
        print("Programa Finalizado.")
        break
    else:
        print("Opción no válida.")

```
