Escribe un programa en Python que aplique el modelo Entrada-Proceso-Salida para
calcular el área de un triángulo (`area = (base * altura) / 2`), usando `input()`,
`float()` y `print()`.


#### Ejercicio 10: Suma acumulada con `while`
Pide números mientras sean distintos de `0` (el `0` termina el ciclo y no se suma).
**Entrada:** `5`, `8`, `2`, `0` → **Salida:** `La suma de los números es: 15`


````python
suma = 0
numero = float(input("Ingrese un número (0 para terminar): "))

while numero != 0:
    suma = suma + numero
    numero = float(input("Ingrese otro número (0 para terminar): "))

print("La suma de los números es:", suma)
````


#### Ejercicio 12: Suma de pares del 1 al 20 con `for` y `continue`
**Salida:** `La suma de los números pares del 1 al 20 es: 110`


````python
suma = 0

for numero in range(1, 21):
    if numero % 2 != 0:
        continue
    
    suma = suma + numero

print("La suma de los números pares del 1 al 20 es:", suma)
````


#### Ejercicio 13: Suma de una lista con función que regresa valor (sin `sum()`)
**Entrada:** `[10, 20, 30, 40]` → **Salida:** `La suma de los elementos es: 100`


````python
def sumar_lista(numeros):
    suma = 0

    for numero in numeros:
        suma = suma + numero

    return suma


numeros = [10, 20, 30, 40]

resultado = sumar_lista(numeros)

print("La suma de los elementos es:", resultado)
````


#### Ejercicio 14: Máximo de una lista con función que regresa valor (sin `max()`)
**Entrada:** `[7, 3, 9, 2]` → **Salida:** `El elemento máximo es: 9`


````python
def encontrar_maximo(numeros):
    mayor = numeros[0]

    for numero in numeros:
        if numero > mayor:
            mayor = numero

    return mayor


numeros = [7, 3, 9, 2]

resultado = encontrar_maximo(numeros)

print("El elemento máximo es:", resultado)
````

#### Ejercicio 15: Matriz y suma de la diagonal principal
**Entrada:** `[[1,2,3],[4,5,6],[7,8,9]]` → **Salida:** `La suma de la diagonal
principal es: 15`


````python
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

suma = 0

for i in range(len(matriz)):
    suma = suma + matriz[i][i]

print("La suma de la diagonal principal es:", suma)
````

#### Ejercicio 16: Promedio con función que no regresa valor
**Entrada:** `[8, 9, 10, 7]` → **Salida:** `El promedio es: 8.50`


````python
def calcular_promedio(calificaciones):
    suma = 0

    for calificacion in calificaciones:
        suma = suma + calificacion

    promedio = suma / len(calificaciones)

    print(f"El promedio es: {promedio:.2f}")


calificaciones = [8, 9, 10, 7]

calcular_promedio(calificaciones)
````

#### Ejercicio 18: Factorial con función recursiva
**Entrada:** `5` → **Salida:** `El factorial de 5 es: 120`


````python
def factorial(numero):
    if numero == 0 or numero == 1:
        return 1
    else:
        return numero * factorial(numero - 1)


numero = int(input("Ingrese un número: "))

resultado = factorial(numero)

print("El factorial de", numero, "es:", resultado)
````

#### Ejercicio 19: Tupla: acceso y suma con función
**Entrada:** `(4, 8, 15, 16, 23, 42)` → **Salida:**
```
El tercer elemento es: 15
La suma de la tupla es: 108
```


````python
def sumar_tupla(numeros):
    suma = 0

    for numero in numeros:
        suma = suma + numero

    return suma


numeros = (4, 8, 15, 16, 23, 42)

print("El tercer elemento es:", numeros[2])

resultado = sumar_tupla(numeros)

print("La suma de la tupla es:", resultado)
````

#### Ejercicio 21: Diccionario de calificaciones y promedios
**Entrada:** `{"Ana": [8, 9, 10], "Beto": [7, 7, 8]}` → **Salida:**
```
Ana: 9.00
Beto: 7.33
```


````python
calificaciones = {
    "Ana": [8, 9, 10],
    "Beto": [7, 7, 8]
}

for nombre in calificaciones:
    notas = calificaciones[nombre]
    suma = 0

    for nota in notas:
        suma = suma + nota

    promedio = suma / len(notas)

    print(f"{nombre}: {promedio:.2f}")
````
