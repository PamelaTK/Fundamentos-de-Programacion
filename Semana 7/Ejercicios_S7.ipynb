### Anexar Un Registro Con El Modo "a"

```python
with open ("bitacora.txt", "w") as file:
    file.write("Registro 1: Sistema Iniciado.\n")
    file.write("Registro 2: Usuario Registrado.\n")


with open ("bitacora.txt" , "a") as file:
    file.write("Registro 3: Operacion Guardada")
```

### Diferencia Entre `'w'` Y `'a'`

```python
file = open("datos.txt" , "w")
file.write("Dato original")
file.close()

with open("datos.txt", "w") as file:
    file.write("Sobrescrito")

with open("datos.txt", "w") as file:
    file.write("Nuevo Valor \n")

with open("datos.txt", "a") as file:
    file.write("Adicional")
```
