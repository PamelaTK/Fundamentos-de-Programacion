```python
import time
import pdb

def pant_carga(usuario):
    print("SISTEMA DE DESCARGA DE AGUA")
    print("Iniciando sistema...")

    for contador in range(3):
        print(f"Cargando{'.' * (contador + 1)}")
        time.sleep(1)

    print("Sistema Listo.\n")

def pant_inicio():
    print("SISTEMA DE DESCARGA DE AGUA")

    usuario = input("Ingrese su nombre o user: ")
    empresa = input("Ingrese el nombre de la empresa: ")
    print("Bienvenido/a", usuario, ". El sistema está listo para la empresa", empresa)
    pant_carga(usuario)

    return usuario, empresa

def capturar_fecha():
    while True:
        print("\nIngrese la fecha de la descarga:")

        try:
            dia = int(input("Ingrese el día de la descarga: "))
            mes = int(input("Ingrese el mes de la descarga: "))
            anio = int(input("Ingrese el año de la descarga: "))
            Fecha = dia, mes, anio
            return Fecha

        except ValueError:
            print("La fecha debe contener solamente números.")

def mostrar_fecha(fecha):
    dia, mes, anio = fecha
    print("Fecha de la descarga:", dia, "/", mes, "/", anio)

def mostrar_menu():
    menu_matriz = [
        ["1", "Registrar Nueva Descarga"],
        ["2", "Consultar Permiso"],
        ["3", "Leer Archivo"],
        ["4", "Crear Archivo"],
        ["5", "Agregar Información a Archivo"],
        ["6", "Mostrar Archivos Disponibles"],
        ["7", "Salir"]
    ]

    print("\nMenú Principal")

    for fila in menu_matriz:
        print(f"{fila[0]}. {fila[1]}")

archs_disps = [
    "Permisos.txt",
    "Descargas.txt",
    "Quimicos.txt",
    "Recibo_Descarga.txt"
]

def prep_archs():
    archs_ini = {
        "Permisos.txt": "Registro de Permisos del Sistema\n",
        "Descargas.txt": "Registro de Descargas de Agua\n",
        "Quimicos.txt": "Registro de Sustancias Quimicas\n",
        "Recibo_Descarga.txt": "Recibos de Descarga de Agua\n"
    }

    for nombre_archivo, contenido in archs_ini.items():
        try:
            with open(nombre_archivo, "x") as file:
                file.write(contenido)

        except FileExistsError:
            pass

        except PermissionError:
            print("No se tienen permisos para crear", nombre_archivo, ".")

def mostrar_archs():
    print("\nArchivos Disponibles")

    for contador in range(len(archs_disps)):
        print(f"{contador + 1}. {archs_disps[contador]}")

def leer_arch():
    mostrar_archs()

    try:
        opcion = int(input("Seleccione el número de archivo que desea abrir: "))

    except ValueError:
        print("Debe ingresar un número.")
        return

    if opcion >= 1 and opcion <= len(archs_disps):
        nombre_arch = archs_disps[opcion - 1]

        try:
            with open(nombre_arch, "r") as file:
                contenido = file.read()
                print("\nContenido del Archivo")
                print(contenido)

        except FileNotFoundError:
            print("No se encontró el archivo.")

        except PermissionError:
            print("No se tienen permisos para leer este archivo.")

    else:
        print("La opción seleccionada no es válida.")

def crear_arch(fecha):
    nomb_arch = input("\nIngrese el nombre del nuevo archivo incluyendo .txt: ")
    contenido = input("Ingrese la información que desea guardar: ")

    try:
        with open(nomb_arch, "w") as file:
            file.write("Fecha: " + str(fecha) + "\n")
            file.write(contenido + "\n")

        print("El archivo se generó correctamente.")
        archs_disps.append(nomb_arch)

    except PermissionError:
        print("No se tiene permiso para crear este archivo.")

def agregar_arch(fecha):
    mostrar_archs()

    try:
        opcion = int(input("\nSeleccione el archivo al que desea agregar información: "))

    except ValueError:
        print("Debe ingresar un número.")
        return

    if opcion >= 1 and opcion <= len(archs_disps):
        nomb_arch = archs_disps[opcion - 1]
        contenido = input("Ingrese la información que desea agregar: ")

        try:
            with open(nomb_arch, "a") as file:
                file.write("Fecha: " + str(fecha) + "\n")
                file.write(contenido + "\n")

            print("La información se agregó correctamente.")

        except FileNotFoundError:
            print("No se encontró el archivo.")

        except PermissionError:
            print("No se tienen permisos para modificar este archivo.")

    else:
        print("La opción seleccionada no es válida.")

def registrar_descarga(fecha, empresa):
    print("\nREGISTRAR NUEVA DESCARGA")

    permiso = input("Número de permiso: ")
    tipo_permiso = input("Tipo de permiso: ")
    acceso = True

    if tipo_permiso != "CONAGUA-01-001" and tipo_permiso != "CONAGUA-01-011":
        print("ACCESO DENEGADO")
        print("El tipo de permiso no es válido.")
        acceso = False

    if acceso == True:
        permiso_valido = input("¿El permiso está en regla? (si/no): ")

        if permiso_valido.lower() != "si":
            print("ACCESO DENEGADO")
            print("El permiso no está en regla.")
            acceso = False

    if acceso == True:
        try:
            with open("Permisos.txt", "a") as file:
                file.write("\n")
                file.write("Fecha: " + str(fecha) + "\n")
                file.write("Empresa: " + empresa + "\n")
                file.write("Número de permiso: " + permiso + "\n")
                file.write("Tipo de permiso: " + tipo_permiso + "\n")

        except FileNotFoundError:
            print("No se encontró Permisos.txt.")

        except PermissionError:
            print("No se tienen permisos para modificar Permisos.txt.")

    if acceso == True:
        try:
            volumen_agua = float(input("Cantidad de agua a descargar en litros: "))

        except ValueError:
            print("La cantidad de agua debe ser numérica.")
            acceso = False

    if acceso == True:
        contiene_quimicos = input("¿El agua contiene sustancias químicas? (si/no): ")

        if contiene_quimicos.lower() == "si":
            try:
                cantidad_quimicos = int(input("¿Cuántas sustancias contiene?: "))

            except ValueError:
                print("Debe ingresar un número.")
                cantidad_quimicos = 0
                acceso = False

            for contador in range(cantidad_quimicos):
                sustancia = input("Nombre de la sustancia: ")

                try:
                    concentracion = float(input("Concentración en ppm: "))

                except ValueError:
                    print("La concentración debe ser numérica.")
                    concentracion = 0

                try:
                    with open("Quimicos.txt", "a") as file:
                        file.write("\n")
                        file.write("Fecha: " + str(fecha) + "\n")
                        file.write("Empresa: " + empresa + "\n")
                        file.write("Sustancia: " + sustancia + "\n")
                        file.write("Concentración: " + str(concentracion) + " ppm\n")

                except FileNotFoundError:
                    print("No se encontró Quimicos.txt.")

                except PermissionError:
                    print("No se tienen permisos para modificar Quimicos.txt.")

            print("ACCESO DENEGADO")
            print("La descarga contiene sustancias químicas.")
            acceso = False

    if acceso == True:
        try:
            with open("Descargas.txt", "a") as file:
                file.write("\n")
                file.write("Fecha: " + str(fecha) + "\n")
                file.write("Empresa: " + empresa + "\n")
                file.write("Número de permiso: " + permiso + "\n")
                file.write("Tipo de permiso: " + tipo_permiso + "\n")
                file.write("Volumen de agua: " + str(volumen_agua) + " litros\n")

            print("\nDatos de la Empresa")
            print("Empresa:", empresa)
            print("Número de permiso:", permiso)
            print("Tipo de permiso:", tipo_permiso)
            print("Volumen de agua:", volumen_agua, "litros")
            print("De acuerdo, tiene autorización para proseguir con su desecho.")

            with open("Recibo_Descarga.txt", "a") as file:
                file.write("\n")
                file.write("RECIBO DE DESCARGA\n")
                file.write("Fecha: " + str(fecha) + "\n")
                file.write("Empresa: " + empresa + "\n")
                file.write("Permiso: " + permiso + "\n")
                file.write("Volumen: " + str(volumen_agua) + " litros\n")

        except FileNotFoundError:
            print("No se encontró el archivo de descarga.")

        except PermissionError:
            print("No se tienen permisos para guardar la descarga.")

def consultar_permiso():
    permiso_buscar = input("\nIngrese el número de permiso que desea consultar: ")
    encontrado = False

    try:
        with open("Permisos.txt", "r") as file:
            contenido = file.read()

        if permiso_buscar in contenido:
            encontrado = True

        if encontrado == True:
            print("El permiso se encuentra registrado.")
        else:
            print("El permiso no se encuentra registrado.")

    except FileNotFoundError:
        print("No se encontró el archivo Permisos.txt.")

    except PermissionError:
        print("No se tienen permisos para leer Permisos.txt.")

def cont_inac(tiempo_transcurrido):
    inactivo = True

    for contador in range(10):
        if tiempo_transcurrido < (contador + 1) * 60:
            inactivo = False
            break

    if inactivo == True:
        print("\nHan pasado 10 minutos sin actividad.")
        continuar = input("¿Desea continuar? (si/no): ")

        if continuar.lower() == "si":
            print("El sistema seguirá en acción.")
            return True
        else:
            print("Regresando a la pantalla de inicio.")
            return False

    return True

def depurar():
    pdb.set_trace()

prep_archs()

reiniciar = True

while reiniciar == True:
    usuario, empresa = pant_inicio()
    fecha = capturar_fecha()
    mostrar_fecha(fecha)

    while True:
        mostrar_menu()
        tiempo_inicio = time.time()

        try:
            opcion = int(input("\nSeleccione una opción: "))

        except ValueError:
            print("Debe ingresar un número válido.")
            continue

        tiempo_final = time.time()
        tiempo_transcurrido = tiempo_final - tiempo_inicio
        continuar = cont_inac(tiempo_transcurrido)

        if continuar == False:
            break

        if tiempo_transcurrido >= 600:
            continue

        if opcion == 1:
            registrar_descarga(fecha, empresa)

        elif opcion == 2:
            consultar_permiso()

        elif opcion == 3:
            leer_arch()

        elif opcion == 4:
            crear_arch(fecha)

        elif opcion == 5:
            agregar_arch(fecha)

        elif opcion == 6:
            mostrar_archs()

        elif opcion == 7:
            print("Saliendo del Sistema -_-")
            reiniciar = False
            break

        else:
            print("La opción seleccionada no es válida.")
```
