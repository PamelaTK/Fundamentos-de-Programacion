```python
distancia = float(input("ingrese la distancia en kilómetros: "))
tiempo = float(input("ingrese el tiempo en horas: "))
print("el dato de distancia es de tipo:", type(distancia))
print("el dato de tiempo es de tipo:", type(tiempo))
velocidad = distancia / tiempo
print("la velocidad promedio es de:", velocidad, "km/h")
```
